**Functions**
* To Reuse a certain block of code just by calling
* Predefined function - Functions that are already there in python
* UserDefined Functions - Functions that we create (users Create)


```python
# Predefined - print(), len(), type(), zip(), range()
# sorted(): Sorts the value in ascending order
number = [5,6,3,4,1,2]
print(sorted(number)) # Ascending order
print(sorted(number,reverse=True)) # Print in Descending order
```

```python
# sum():
number = [5,6,3,4,1,2]
print(sum(number))
```

```python
# max():
number = [5,6,3,4,1,2]
print(max(number))
```

```python
# min()
number = [5,6,3,4,1,2]
print(min(number))
```

```python
# round()
x = 2.6544543
print(round(x,1))
```

```python
# abs(): Absoulute
x = -50
print(abs(x))
```

```python
print(abs(-50)+abs(50))
```

**User Defined Funtion**

```python
# keyword alert: def
# Non parameterized
def displayName():
  print('Hello I am Aryan')
```

```python
# call the funtion
displayName()
displayName()
displayName()
displayName()
```

```python
# Parameteruzed function
def addition(x,y,z):
  print(f'Sum is {x+y+z}')
```

```python
addition(5,6,7)
```

```python
# Default parameter
def multi(x,y=1):
  print(f'Multiplication of {x} and {y}:{x*y}')
```

```python
multi(10,10)
```

**Lambda Function**
- One statement function

```python
#keyword : Lambda
# defined the func -> Lambda
add = lambda x,y : print(f'{x} + {y} = {x+y}')

# call the function
add(2,3)
```
